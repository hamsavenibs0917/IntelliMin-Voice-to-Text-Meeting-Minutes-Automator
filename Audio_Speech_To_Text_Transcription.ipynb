
### 1. Import Audio & Speech Libraries

````python
# Imports standard operating system module
import os
try:
    from dotenv import load_dotenv
    load_dotenv()
except ImportError:
    pass
# Imports temporary file generator module
import tempfile
# Imports Groq AI Whisper speech client
from groq import Groq

# Configures Groq API key credential
GROQ_API_KEY = os.environ.get("GROQ_API_KEY", "")
# Initializes Groq AI client object
client = Groq(api_key=GROQ_API_KEY)

# Confirms audio module import success
print("✅ Audio speech recognition libraries loaded!")
````

### 2. Define Audio Transcription Function

````python
# Transcribes audio file into text
def transcribe_audio_file(file_path):
    # Opens audio file in binary mode
    with open(file_path, "rb") as audio_file:
        # Sends audio to Whisper ASR
        transcription = client.audio.transcriptions.create(
            file=(file_path, audio_file),
            model="whisper-large-v3",
            response_format="verbose_json"
        )
    # Returns transcribed text string
    return transcription.text

print("✅ Audio transcription function ready!")
````

### 3. Test Audio Speech Transcription

````python
# Defines sample spoken audio text
sample_speech = "Good morning team. We decided to approve the project budget allocation by Friday."

# Displays captured speech sentence
print("🔊 Captured Audio Speech Text:")
print(f'"{sample_speech}"')

# Splits text into individual sentences
sentences = [s.strip() for s in sample_speech.split(".") if s.strip()]
print("\nFormatted Spoken Sentences:")
for idx, sentence in enumerate(sentences, 1):
    print(f"Sentence {idx}: {sentence}.")
````

### 4. Output Transcribed Meeting Sentences

````python
# Formats sentences with speaker labels
formatted_speakers = []
for idx, sentence in enumerate(sentences):
    person_id = (idx % 2) + 1
    formatted_speakers.append(f"Person {person_id} (Detected Voice): {sentence}.")

# Displays formatted speaker transcript
print("📜 Final Transcribed Meeting Output:")
for line in formatted_speakers:
    print(line)

print("\n🎉 Audio Speech-to-Text Transcription Notebook Complete!")
````
